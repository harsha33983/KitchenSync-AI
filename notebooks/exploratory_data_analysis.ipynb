# KitchenSync AI: Exploratory Data Analysis & Feature Investigation
## Food Delivery & Q-Commerce Real-Time Kitchen Prep & Rider Dispatch Estimator

This notebook performs exploratory data analysis on the real-world food delivery operations dataset.
We examine preparation durations, delivery distances, traffic conditions, and kitchen workload.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 6)

# Load processed dataset
df = pd.read_csv('../data/processed/food_delivery_features.csv')
print(f'Loaded dataset: {df.shape[0]:,} rows, {df.shape[1]} columns')
df.head()
```

### 1. Target Distributions: Food Preparation Time & Rider Transit Time

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['actual_prep_time'], kde=True, ax=axes[0], color='#2b5c8f')
axes[0].set_title('Kitchen Preparation Time Distribution')
axes[0].axvline(df['actual_prep_time'].median(), color='red', linestyle='--', label=f'P50 Median: {df["actual_prep_time"].median():.1f}m')
axes[0].axvline(df['actual_prep_time'].quantile(0.90), color='orange', linestyle='--', label=f'P90: {df["actual_prep_time"].quantile(0.90):.1f}m')
axes[0].legend()

sns.histplot(df['actual_rider_travel_time'], kde=True, ax=axes[1], color='#287d55')
axes[1].set_title('Rider Delivery Travel Time Distribution')
axes[1].axvline(df['actual_rider_travel_time'].median(), color='red', linestyle='--', label=f'Median: {df["actual_rider_travel_time"].median():.1f}m')
axes[1].legend()
plt.show()
```

### 2. Operational Impact: Kitchen Queue Depth vs Preparation Time

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x='kitchen_queue_depth', y='actual_prep_time', data=df, palette='Reds')
plt.title('Kitchen Queue Congestion vs Food Preparation Duration')
plt.xlabel('Active Kitchen Orders in Queue')
plt.ylabel('Preparation Duration (min)')
plt.show()
```

### 3. Traffic Conditions vs Rider Transit Travel Time

```python
plt.figure(figsize=(9, 5))
sns.boxplot(x='traffic_level', y='actual_rider_travel_time', data=df, order=['Low', 'Medium', 'High', 'Jam'], palette='Oranges')
plt.title('Impact of Road Traffic Density on Transit Duration')
plt.xlabel('Traffic Density')
plt.ylabel('Rider Travel Time (min)')
plt.show()
```

### 4. Correlation Heatmap across Engineered Features

```python
num_cols = ['actual_prep_time', 'actual_rider_travel_time', 'item_count', 'kitchen_queue_depth', 'distance_km', 'store_load', 'order_complexity']
plt.figure(figsize=(9, 7))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Operational Feature Correlation Matrix')
plt.show()
```
